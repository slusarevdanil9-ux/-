```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
from collections import Counter

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'

df1 = pd.read_csv('Отзывы-Еда.csv', encoding='utf-8', encoding_errors='replace')

bad_words = [
    'магазин', 'супермаркет', 'бар', 'диско', 'батут', 'парк', 'клуб', 'алкомаркет',
    'Пекарн', 'маркет', 'кондитерская', 'Автокафе', 'Гастрономический бутик',
    'Пункт самовывоза', 'Продуктовый дискаунтер', 'лавка', 'Павильон',
    'Официальный представитель Железнодорожник, АО', 'Алкогольные напитки', 'Бытовая химия'
]
pattern = '|'.join([re.escape(w.strip()) for w in bad_words])

df = df1[~df1['Specialization'].str.contains(pattern, case=False, na=False)].copy()

for col in ['Brand', 'Adress', 'Comment', 'Specialization']:
    if col in df.columns:
        df[col] = df[col].astype(str).str.replace('\xa0', ' ').str.replace('\u200b', '').str.strip()

df['Score'] = pd.to_numeric(df['Score'], errors='coerce')
df['Date'] = pd.to_datetime(df['Date'], format='%d.%m.%Y', errors='coerce')
df['Year'] = df['Date'].dt.year

df = df.drop_duplicates(subset=['Brand', 'Date', 'Score', 'Comment'])
#Считываем данные. Удаляем дубликаты. Удаляем заведения не являющиеся нашими конкурентами, которые содержат "плохие слова" в специализации. Выбранная компания Sushi Holl
```

```python

def extract_city(address):
    if pd.isna(address) or address == 'nan':
        return 'Не указан'
    clean_addr = re.sub(r'\d+\s*филиал.*', '', address).strip()
    parts = [p.strip() for p in clean_addr.split(',') if p.strip()]
    return parts[-1] if parts else 'Не указан'

df['City'] = df['Adress'].apply(extract_city)

target_brand = 'Sushi Holl'
target_cities = df[df['Brand'] == target_brand]['City'].unique()
print(f"Города присутствия {target_brand}: {list(target_cities)}")

df = df[df['City'].isin(target_cities)].copy()

point_avg_score = df.groupby(['Brand', 'Adress'])['Score'].transform('mean')

df = df[
    (point_avg_score >= 3.3) &
    (df['Specialization'].str.contains('доставк', case=False, na=False))
].reset_index(drop=True)

print(f"\nПосле фильтрации (рейтинг >= 3.3 и только доставка):")
print(f"Всего компаний в выборке: {df['Brand'].nunique()}")
print(f"Всего отзывов в выборке: {len(df)}")

summary = df.groupby(['City', 'Brand', 'Specialization', 'Adress']).agg(
    Отзывов=('Score', 'count'),
    Средний_балл=('Score', lambda x: round(x.mean(), 2)),
    Доля_5_звезд_проц=('Score', lambda x: round((x == 5).mean() * 100, 1)),
    Доля_негатива_1_2_проц=('Score', lambda x: round((x <= 2).mean() * 100, 1)),
    Активность_с=('Date', 'min'),
    Активность_по=('Date', 'max')
).reset_index().sort_values(by=['City', 'Отзывов'], ascending=[True, False])

print("\n=== ТОЧКИ ДОСТАВКИ С РЕЙТИНГОМ >= 3.3 В ГОРОДАХ ПРИСУТСТВИЯ SUSHI HOLL ===")
print(summary.to_string(index=False))
#Берём только те компании у которых есть филиалы в одном городе с Sushi Holl. Также для уменьшения выборки удаляем заведения с рейтингом ниже 3.3. Можно удалить так как они составляют малую конкуренцию.
```

```python
pain_points = {
    'Долгая доставка / опоздания': r'опозда|долго|ждал|ждать|задерж|время ожидания|вовремя|привезли через',
    'Холодная еда': r'холодн|остывш|еле тепл|чуть тепл|замерз',
    'Мало начинки / много риса / размер': r'мало|один рис|тонк|прозрачн|пожалели|маленьк|недовес|экономи',
    'Качество / свежесть / отравление': r'несвеж|тухл|воня|запах|отрав|живот|сыр[аоы]|волос|таракан|грязн|пересол|подгор',
    'Сервис / грубость / ошибки': r'хам|груб|не извинил|забыли|не положили|перепутал|не берут трубку|оператор|администратор',
    'Проблемы с акциями / сайтом / ценой': r'акци|промокод|бонус|скидк|сайт|приложен|дорог|ценник|сдач'
}

neg_df = df[df['Score'] <= 2].copy()

results = []
for brand, group in neg_df.groupby('Brand'):
    total_neg = len(group)
    row = {'Brand': brand, 'Негативных_отзывов': total_neg}
    for issue, pattern in pain_points.items():
        count = group['Comment'].str.lower().str.contains(pattern, regex=True, na=False).sum()
        row[issue] = round((count / total_neg) * 100, 1)
    results.append(row)

pain_matrix = pd.DataFrame(results).set_index('Brand')

print("\n=== ДОЛЯ НЕГАТИВНЫХ ОТЗЫВОВ С УПОМИНАНИЕМ ПРОБЛЕМЫ (%) ===")
print(pain_matrix.to_string())

plt.figure(figsize=(12, 6))
sns.heatmap(pain_matrix.drop(columns=['Негативных_отзывов']), annot=True, cmap='Reds', fmt='.1f')
plt.title('Тепловая карта проблем в негативных отзывах (% от всех оценок 1 и 2)', fontsize=13, fontweight='bold')
plt.ylabel('')
plt.xticks(rotation=25, ha='right')
plt.tight_layout()
plt.show()
#Выявляем пробелы описанные в отзывах для всех компаний и составляем по ним тепловую карту.
#Сдесь уже можно увидеть, что самая часто встречающаяся причина негативных отзывов для нашей компании это проблемы с доставкой.
```

```python
stop_words = {
    'и', 'в', 'во', 'не', 'что', 'он', 'на', 'я', 'с', 'со', 'как', 'а', 'то', 'все', 'она',
    'так', 'его', 'но', 'да', 'ты', 'к', 'у', 'же', 'вы', 'за', 'бы', 'по', 'только', 'ее',
    'мне', 'было', 'вот', 'от', 'меня', 'еще', 'нет', 'о', 'из', 'ему', 'теперь', 'когда',
    'даже', 'ну', 'вдруг', 'ли', 'если', 'уже', 'или', 'ни', 'быть', 'был', 'него', 'до',
    'вас', 'нибудь', 'опять', 'уж', 'вам', 'ведь', 'там', 'потом', 'себя', 'ничего', 'ей',
    'может', 'они', 'тут', 'где', 'есть', 'надо', 'ней', 'для', 'мы', 'тебя', 'их', 'чем',
    'была', 'сам', 'чтоб', 'без', 'будто', 'чего', 'раз', 'тоже', 'себе', 'под', 'будет',
    'ж', 'тогда', 'кто', 'этот', 'того', 'потому', 'этого', 'какой', 'совсем', 'ним', 'здесь',
    'этом', 'один', 'почти', 'мой', 'тем', 'чтобы', 'нее', 'сейчас', 'были', 'куда', 'зачем',
    'всех', 'никогда', 'можно', 'при', 'наконец', 'два', 'об', 'другой', 'хоть', 'после',
    'над', 'больше', 'тот', 'через', 'эти', 'нас', 'про', 'всего', 'них', 'какая', 'много',
    'разве', 'три', 'эту', 'моя', 'впрочем', 'хорошо', 'свою', 'этой', 'перед', 'иногда',
    'лучше', 'чуть', 'том', 'нельзя', 'такой', 'им', 'более', 'всегда', 'конечно', 'всю', 'между',
    'очень', 'просто', 'это', 'все', 'всё', 'очень', 'вообще', 'просто'
}

def get_top_words(text_series, n=15):
    words = []
    for text in text_series.dropna():
        tokens = re.findall(r'[а-яё]{4,}', text.lower())
        words.extend([w for w in tokens if w not in stop_words])
    return Counter(words).most_common(n)

target = df[df['Brand'] == 'Sushi Holl']

print("\n=== ТОП-15 СЛОВ В ПОЗИТИВНЫХ ОТЗЫВАХ SUSHI HOLL (Оценка 5) ===")
for word, count in get_top_words(target[target['Score'] == 5]['Comment']):
    print(f"{word}: {count}")

print("\n=== ТОП-15 СЛОВ В НЕГАТИВНЫХ ОТЗЫВАХ SUSHI HOLL (Оценки 1-2) ===")
for word, count in get_top_words(target[target['Score'] <= 2]['Comment']):
    print(f"{word}: {count}")

#Выделяем 15 самых часто встречающихся слов в негативных и положительных отзывах для Sushi Holl
#Проанализировав их понимаем, что жалобы поступают в основном на доставку, в то время как сами роллы хвалят.
```

```python
def get_complaint_category(row):
    if row['Score'] > 3:
        return 'Без жалоб'

    text = str(row['Comment']).lower()

    if any(w in text for w in ['доставк', 'опозда', 'долго', 'час', 'курьер', 'холодн', 'осты', 'райо', 'везут', 'везли', 'ждать', 'ждали']):
        return 'Доставка и логистика'
    elif any(w in text for w in ['рис', 'рыб', 'начинк', 'развал', 'сух', 'невкусн', 'пицц', 'пересол', 'сыр', 'волос', 'тухл', 'несвеж', 'отрав', 'живот']):
        return 'Качество еды и кухня'
    elif any(w in text for w in ['оператор', 'подарок', 'подароч', 'забыл', 'положил', 'не долож', 'трубк', 'отмен', 'хам', 'груб', 'менеджер', 'администратор', 'прибор', 'палочк']):
        return 'Сервис и сборка заказа'
    elif any(w in text for w in ['акци', 'промокод', 'скидк', 'сайт', 'оплат', 'приложен', 'бонус', 'чек', 'цен']):
        return 'Акции, цены и сайт'
    else:
        return 'Прочее / Общий негатив'

def get_mentioned_dish(comment):
    text = str(comment).lower()
    dishes = []

    if any(w in text for w in ['ролл', 'суши', 'филадельф', 'сет', 'темпур', 'калифорн', 'сяке', 'лава']):
        dishes.append('Роллы и суши')
    if any(w in text for w in ['пицц', 'пепперон']):
        dishes.append('Пицца')
    if any(w in text for w in ['том ям', 'то ям', 'вок', 'wok', 'лапш', 'суп', 'рамен', 'фо бо', 'удон']):
        dishes.append('Супы и WOK')
    if any(w in text for w in ['крыл', 'куриц', 'курочк', 'чикен', 'ножк', 'филе']):
        dishes.append('Курица / Крылья')
    if any(w in text for w in ['пирог', 'бурек', 'курник', 'киш', 'шашлык', 'люля', 'хинкал', 'хачапур']):
        dishes.append('Пироги и Кавказская кухня')

    if not dishes:
        return 'Не указано'
    elif len(dishes) == 1:
        return dishes
    else:
        return 'Несколько блюд'

df['Категория_жалобы'] = df.apply(get_complaint_category, axis=1)
df['Упомянутое_блюдо'] = df['Comment'].apply(get_mentioned_dish)



df.to_csv("output1.csv")
#Создаем новые столбцы. Они нужны для дальнейшей работы.
```
